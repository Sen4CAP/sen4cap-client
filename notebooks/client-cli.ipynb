# Client CLI Usage

This notebook runs the `sen4cap-client` executable from the notebook environment.
Configure and log in in a terminal first:

```bash
sen4cap-client configure
sen4cap-client login
```

Use your administrator's service URLs; the defaults point to localhost.
Interactive prompts need a terminal. Noninteractive configuration is possible
when all required options are supplied. See [configuration](https://Sen4CAP.github.io/sen4cap-client/configuration/).

These cells make live service calls. Process `218` is an example; inspect the
process description and edit the generated request before submitting a job.

````python
!sen4cap-client --help
````

````python
!sen4cap-client list-processes
````

````python
process_id = "218"
!sen4cap-client get-process {process_id}
````

Generate a request template, then save the printed JSON as UTF-8 `request.json` and edit it before continuing. Avoid plain `>` redirection in Windows PowerShell 5.1, which writes UTF-16.

````python
!sen4cap-client create-request {process_id} --format json
````

Validate the request structure locally. This does not check its input values against the remote process schema.

````python
!sen4cap-client validate-request --request request.json
````

Submit the edited request. The command returns immediately with a job ID; it does not wait for processing.

````python
!sen4cap-client execute-process --request request.json
````

````python
!sen4cap-client list-jobs
````

````python
job_id = input("Job ID returned by execute-process: ").strip()
!sen4cap-client get-job {job_id}
````

Run the next cell once the job status is `successful`. It retrieves output values/links, not the raster data. Use the Python API to open a raster.

````python
!sen4cap-client get-job-results {job_id}
````

Other commands include `dismiss-job JOB_ID` to cancel a running job or delete a
finished one, and `show-app` to open the GUI (run it in a terminal and stop it
with Ctrl+C). `logout` removes locally saved credentials. Use each command's
`--help` for its options. See the [CLI reference](https://Sen4CAP.github.io/sen4cap-client/cli/).
