# Client CLI Usage

The following is a step-by-step guide that demonstrates how to use the Sen4CAP CLI.

Similar to the Python API client, the CLI client provides the following capabilities:

- `sen4cap-client list-processes`: list existing processes,
- `sen4cap-client get-process`: get the details about a process,
- `sen4cap-client validate-request`: validate a given process execution request,
- `sen4cap-client execute-process`: execute a given process execution request,
- `sen4cap-client list-jobs`: list the jobs resulting from process executions, and finally
- `sen4cap-client get-job`: get a job's details,
- `sen4cap-client get-job-results`: get a successful job's results, and finally
- `sen4cap-client dismiss-job`: cancel a successful job.
- `sen4cap-client show-app`: show the client app in a browser.

To use the CLI you'll need to configure it first by entering in another terminal (the command does not work in Jupyter notebooks):

```bash
sen4cap-client configure
```

After configuration you can use the `login` and `logout` commands.
